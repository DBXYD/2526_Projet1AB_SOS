```python

clear all ;
close all ;
clc;
format short g ;

fa=1053
fe=4000
f0=1000
sa=1
nu0 = fa/fe ;
Dobs = 0.5 ;
Nobs = 0:Dobs*fe-1 ;
sig = sin(2*pi*nu0*Nobs) ;
sound(sig,fe) ;

% Appel de la fonction DTFTtp pour calculer la transformée de Fourier
[Nu, S] = DTFTtp(sig);

%p3:

% Calcul de l'échelle temporelle
t = (0:length(sig)-1) / fe;

% Calcul de l'échelle fréquentielle
f = linspace(0, fe/2, length(Nu)); % Échelle linéaire
f_dB = 20*log10(abs(S)); % Échelle dB

% Création de la figure
figure;

subplot(3,1,1);
plot(t, sig);
%xlabel('Temps (s)');
%ylabel('Amplitude');
%title('Évolution temporelle de la séquence');
xlim([0 0.1]);

subplot(3,1,2);
plot(Nu, abs(S));
xlabel('Fréquence (Hz)');
ylabel('Magnitude');
title('Évolution fréquentielle de la séquence');

subplot(3,1,3);
plot(Nu,20*log(abs(S)) );
xlabel('Fréquence (Hz)');
ylabel('Magnitude (dB)');
title('Évolution fréquentielle (échelle logarithmique)');


% p4. Calculer la séquence sig2 en tant que sortie du filtre de fonction transfert :
% H(z) = 1 - 2z^(-1) + z^(-2) à la séquence sig (cf. II.2.4). Calculer sa TF(T)D Sig2 puis superposer
% ses tracés à ceux de sig et Sig sur la figure précédente.

% Définition du filtre
b = [1, -2, 1];
a = 1;

% Filtrage de la séquence sig
sig2 = filter(b, a, sig);

% Calcul de la TF(T)D de sig2
[Nu2, S2] = DTFTtp(sig2);

% Superposition des tracés
hold on;
subplot(3,1,1);
plot(t, sig2, '--');
legend('sig', 'sig2');
subplot(3,1,2);
plot(Nu2, abs(S2), '--');
legend('Sig', 'Sig2');
subplot(3,1,3);
plot(Nu2, 20*log10(abs(S2)), '--');
legend('Sig', 'Sig2');
hold off;

% p5. Calculer les réponses impulsionnelle h et en fréquence H du filtre. Comparer le résultat obtenu
% à celui de impz. Sur une nouvelle figure, tracer h puis H en module et phase sur 3 chronogrammes
% alignés verticalement. Justifier les représentations de sig2.

% Calcul de la réponse impulsionnelle et en fréquence du filtre
[h, t_imp] = impz(b, a);
[H, Nu_H] = freqz(b, a, length(Nu2), fe);

% Nouvelle figure
figure;
subplot(3,1,1);
stem(t_imp, h);
xlabel('Temps (échantillons)');
ylabel('Amplitude');
title('Réponse impulsionnelle du filtre');

subplot(3,1,2);
plot(Nu_H, abs(H));
xlabel('Fréquence (Hz)');
ylabel('Magnitude');
title('Réponse en fréquence du filtre (module)');

subplot(3,1,3);
plot(Nu_H, angle(H));
xlabel('Fréquence (Hz)');
ylabel('Phase (radians)');
title('Réponse en fréquence du filtre (phase)');

% Justification des représentations de sig2 : sig2 est la sortie du filtre à réponse impulsionnelle finie
% H(z) = 1 - 2z^(-1) + z^(-2), donc son comportement fréquentiel est modifié par ce filtre.
```
